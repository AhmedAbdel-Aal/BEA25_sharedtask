```python
sentences = []
labels = []
iids = []

for example in dev_data:
    for tutor_name in example['tutor_responses']:
        response = example['tutor_responses'][tutor_name]['response']
        sentences.append(response)
        labels.append(tutor_name)
        iids.append(example['conversation_id'])

len(sentences), len(labels)
```
