```python
dev_truth = load_json("data/source/mrbench_v3_devset.json")
```

```python
tutors=  list(dev_truth[0]['tutor_responses'].keys())
tutors
```

```python
for t in tutors:
    yes_responses = []
    no_responses = []
    some_responses = []
    for i in dev_truth:
        gpt_response = i["tutor_responses"][t]['response']
        mistake = i["tutor_responses"][t]['annotation']['Mistake_Identification']

        if mistake == "Yes":
            yes_responses.append(gpt_response)
        elif mistake == "No":
            no_responses.append(gpt_response)
        else:
            some_responses.append(gpt_response)
    # write the lists in a file
    with open(f"tutor_anaylsis/{t}_yes_responses.txt", "w") as f:
        for item in yes_responses:
            f.write("%s\n" % item)
    with open(f"tutor_anaylsis/{t}_no_responses.txt", "w") as f:
        for item in no_responses:
            f.write("%s\n" % item)
    with open(f"tutor_anaylsis/{t}_some_responses.txt", "w") as f:
        for item in some_responses:
            f.write("%s\n" % item)
```
